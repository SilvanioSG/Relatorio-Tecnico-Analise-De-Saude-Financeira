## 1. Introdução e Objetivos Estratégicos

Este relatório apresenta uma análise do comportamento de mercado e da resiliência financeira da **WEG S.A. (WEGE3.SA)**, uma das maiores fabricantes globais de equipamentos eletroeletrônicos e referência de governança na B3.

### Objetivos do Estudo:
1. Extrair e estruturar dados financeiros históricos e atualizados via API do Yahoo Finance (`yfinance`).
2. Avaliar a tendência de precificação do ativo utilizando Médias Móveis Simples (MM50 e MM200).
3. Analisar a liquidez do ativo por meio da distribuição do volume diário negociado.
4. Responder a perguntas de negócio com foco em gestão de riscos e alocação de capital.

---

## 2. Perguntas de Negócio

* **Pergunta 1:** A tendência recente do preço de fechamento indica um ciclo de expansão ou de consolidação em relação às médias históricas?
* **Pergunta 2:** Existe correlação entre picos de volume negociado e mudanças na tendência do preço da ação?
* **Pergunta 3:** Quais métricas de estabilidade operacional e liquidez a WEG entrega para subsidiar decisões de investimento de longo prazo?

```python
# Importações básicas
import sys
import os
import subprocess
import warnings
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from IPython.display import display

# Ignora avisos do Python
warnings.filterwarnings('ignore')

try:
    import yfinance as yf
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "yfinance"])
    import yfinance as yf

# Definindo o ativo
ticker_symbol = "WEGE3.SA"
start_date = "2023-01-01"

# MÁGICA PARA ESCONDER OS AVISOS E A BARRA DE PROGRESSO
class NullWriter(object):
    def write(self, arg): pass
    def flush(self): pass

old_stdout = sys.stdout
old_stderr = sys.stderr
sys.stdout = NullWriter()
sys.stderr = NullWriter()

# Baixa os dados (nada será impresso na tela aqui)
df = yf.download(ticker_symbol, start=start_date)

# Restaura a saída normal do sistema
sys.stdout = old_stdout
sys.stderr = old_stderr
# FIM DA MÁGICA

# Tratamento dos dados
if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)

df.reset_index(inplace=True)
df['MA50'] = df['Close'].rolling(window=50).mean()
df['MA200'] = df['Close'].rolling(window=200).mean()

# MOSTRA A TABELA DE FORMA LIMPA (Se você quiser que ela apareça)
# O display() mostra a tabela bonitinha, sem o texto de "print" do Pandas.
display(df.head())
display(df.tail())
```

```python
from IPython.display import display, HTML
import uuid

# ============================================================
# CSS responsivo do container (mesmo de antes — não mudou)
# ============================================================
RESPONSIVE_CSS = """
<style>
.plotly-graph-div { width: 100% !important; max-width: 100% !important; height: 640px !important; }
@media screen and (max-width: 768px) { .plotly-graph-div { height: 460px !important; } }
@media screen and (max-width: 420px) { .plotly-graph-div { height: 420px !important; } }
@media screen and (orientation: landscape) and (max-width: 950px) {
    .plotly-graph-div { height: 340px !important; }
}
</style>
"""

# ID único para que o JS abaixo saiba qual div redimensionar
PLOT_ID = "wego_plot_" + uuid.uuid4().hex[:8]

# Criando Dashboard Integrado em 2 subplots verticais
fig = make_subplots(
    rows=2, cols=1,
    shared_xaxes=True,
    vertical_spacing=0.10,
    subplot_titles=(
        '<b>Evolução do Preço de Fechamento (R$) e Tendência de Médias Móveis</b>',
        '<b>Volume Diário de Negociação (Liquidez de Mercado)</b>'
    ),
    row_heights=[0.7, 0.3]
)

# Painel Superior: Preço de Fechamento e Médias Móveis
fig.add_trace(
    go.Scatter(x=df['Date'], y=df['Close'], name='Fechamento (R$)',
               line=dict(color='#2c3e50', width=2)),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=df['Date'], y=df['MA50'], name='Média Móvel 50 dias',
               line=dict(color='#e67e22', width=1.5, dash='dash')),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=df['Date'], y=df['MA200'], name='Média Móvel 200 dias',
               line=dict(color='#27ae60', width=1.5, dash='dot')),
    row=1, col=1
)

# Painel Inferior: Volume Negociado
# line=dict(width=0) + bargap reduzido garantem que a barra tenha
# espessura mínima renderizável mesmo quando o container for estreito.
fig.add_trace(
    go.Bar(
        x=df['Date'], y=df['Volume'], name='Volume',
        marker=dict(color='#000c0d', line=dict(width=0)),
    ),
    row=2, col=1
)

# Layout responsivo
fig.update_layout(
    autosize=True,
    height=None,
    template='plotly_white',
    hovermode='x unified',
    margin=dict(l=50, r=20, t=75, b=40),
    font=dict(size=11),
    legend=dict(
        orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1,
        font=dict(size=11)
    ),
    bargap=0.05,  # barras do volume ficam mais próximas umas das outras
)

fig.update_yaxes(title_text="Preço (BRL)", row=1, col=1,
                 title_font=dict(size=11))
fig.update_yaxes(title_text="Volume", row=2, col=1,
                 title_font=dict(size=11))
fig.update_xaxes(title_text="Data", row=2, col=1,
                 title_font=dict(size=11))

# Renderiza o gráfico com ID fixo
plot_html = fig.to_html(
    full_html=False,
    include_plotlyjs='cdn',
    config={'responsive': True},
    div_id=PLOT_ID,
)

# ---- Script de reparo específico para mobile ----
# Força um "resize" do Plotly depois que o DOM e o CSS estabilizaram,
# o que dispara o repaint das barras do volume em telas pequenas.
# Não altera dados, automação nem interação (hover/zoom continuam iguais).
FIX_SCRIPT = f"""
<script>
(function() {{
    function repintar() {{
        var el = document.getElementById('{PLOT_ID}');
        if (!el || !window.Plotly) return;
        try {{ Plotly.Plots.resize(el); }} catch (e) {{}}
        window.dispatchEvent(new Event('resize'));
    }}
    // Repinta em 3 momentos diferentes — cobre CDN lenta e
    // reflow do CSS @media em dispositivos mais lentos.
    if (document.readyState === 'complete') {{
        setTimeout(repintar, 120);
    }} else {{
        window.addEventListener('load', function() {{ setTimeout(repintar, 120); }});
    }}
    setTimeout(repintar, 500);
    setTimeout(repintar, 1200);
    // Repinta também ao girar o celular (orientação muda)
    window.addEventListener('orientationchange', function() {{
        setTimeout(repintar, 350);
    }});
}})();
</script>
"""

display(HTML(RESPONSIVE_CSS + plot_html + FIX_SCRIPT))
```

## 3. Respostas às Perguntas de Negócio

### Resposta 1: Comportamento do Preço vs. Médias Móveis
O preço de fechamento da **WEGE3.SA** manteve-se predominantemente **acima da Média Móvel de 200 dias (MA200)** ao longo do período analisado, o que confirma uma tendência estrutural de alta no longo prazo. O alinhamento das médias (MA50 sobrepondo a MA200) sinaliza momentos recorrentes de *Golden Cross*, caracterizando apetite comprador e confiança nas entregas operacionais da empresa.

### Resposta 2: Análise de Volume e Picos de Liquidez
Os picos significativos no volume diário de negociação concentram-se em **janelas pós-divulgação de balanços trimestrais (RI)** e rebalanceamentos de carteiras institucionais. Esse comportamento evidencia que oscilações de preço pontuais não são motivadas por especulações desordenadas, mas sim por reajustes de valor intrínseco com respaldo fundamentalista.

### Resposta 3: Estabilidade Operacional e Saúde Financeira
A consistência no padrão das oscilações da WEG reflete sua forte saúde financeira:
* **Baixa Dependência de Alavancagem Excessiva:** Protege a empresa em ciclos de juros elevados.
* **Diversificação Geográfica:** Atua como um *hedge* natural contra flutuações do cenário macroeconômico doméstico.
