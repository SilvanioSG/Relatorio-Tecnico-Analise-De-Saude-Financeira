## 1. Introdução e Objetivos Estratégicos

Este relatório apresenta uma análise do comportamento de mercado e da resiliência financeira da **WEG S.A. (WEGE3.SA)**, uma das maiores fabricantes globais de equipamentos eletroeletrônicos e referência de governança na B3.

### Objetivos do Estudo:
1. Extrair e estruturar dados financeiros históricos e atualizados via API do Yahoo Finance (`yfinance`).
2. Avaliar a tendência de precificação do ativo utilizando Médias Móveis Simples (MM50 e MM200).
3. Analisar a liquidez do ativo por meio da distribuição do volume diário negociado.
4. Responder a perguntas de negócio com foco em gestão de riscos e alocação de capital.

## 2. Perguntas de Negócio

* **Pergunta 1:** A tendência recente do preço de fechamento indica um ciclo de expansão ou de consolidação em relação às médias históricas?
* **Pergunta 2:** Existe correlação entre picos de volume negociado e mudanças na tendência do preço da ação?
* **Pergunta 3:** Quais métricas de estabilidade operacional e liquidez a WEG entrega para subsidiar decisões de investimento de longo prazo?

```python
# Garante a instalação das bibliotecas caso não estejam presentes no ambiente do Jupyter
import sys
import subprocess

def install(package):
    subprocess.check_call([sys.executable, "-m", "pip", "install", package])

# Tentativa de importação e instalação automática se necessário
try:
    import yfinance as yf
except ImportError:
    print("Instalando yfinance...")
    install("yfinance")
    import yfinance as yf

try:
    import plotly
except ImportError:
    print("Instalando plotly...")
    install("plotly")
    import plotly

import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Definindo o ativo e o período de análise
ticker_symbol = "WEGE3.SA"
start_date = "2023-01-01"

print(f"Baixando dados financeiros para {ticker_symbol}...")
df = yf.download(ticker_symbol, start=start_date)

# Tratamento para garantir estrutura de colunas simples (Flat index)
if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)

df.reset_index(inplace=True)

# Engenharia de Variáveis (Médias Móveis)
df['MA50'] = df['Close'].rolling(window=50).mean()
df['MA200'] = df['Close'].rolling(window=200).mean()

# Exibição das últimas linhas tratadas
df.tail()
```

```python
# Criando Dashboard Integrado em 2 subplots verticais
fig = make_subplots(
    rows=2, cols=1, 
    shared_xaxes=True,
    vertical_spacing=0.08,
    subplot_titles=(
        '<b>Evolução do Preço de Fechamento (R$) e Tendência de Médias Móveis</b>', 
        '<b>Volume Diário de Negociação (Liquidez de Mercado)</b>'
    ),
    row_heights=[0.7, 0.3]
)

# Painel Superior: Preço de Fechamento e Médias Móveis
fig.add_trace(
    go.Scatter(x=df['Date'], y=df['Close'], name='Fechamento (R$)', line=dict(color='#2c3e50', width=2)),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=df['Date'], y=df['MA50'], name='Média Móvel 50 dias', line=dict(color='#e67e22', width=1.5, dash='dash')),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=df['Date'], y=df['MA200'], name='Média Móvel 200 dias', line=dict(color='#27ae60', width=1.5, dash='dot')),
    row=1, col=1
)

# Painel Inferior: Volume Negociado
fig.add_trace(
    go.Bar(x=df['Date'], y=df['Volume'], name='Volume', marker_color='#95a5a6'),
    row=2, col=1
)

# Personalização e Layout Profissional
fig.update_layout(
    height=650,
    template='plotly_white',
    hovermode='x unified',
    margin=dict(l=40, r=40, t=60, b=40),
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)

fig.update_yaxes(title_text="Preço (BRL)", row=1, col=1)
fig.update_yaxes(title_text="Volume", row=2, col=1)
fig.update_xaxes(title_text="Data", row=2, col=1)

fig.show()
```

## 3. Respostas às Perguntas de Negócio

### Resposta 1: Comportamento do Preço vs. Médias Móveis
O preço de fechamento da **WEGE3.SA** manteve-se predominantemente **acima da Média Móvel de 200 dias (MA200)** ao longo do período analisado, o que confirma uma tendência estrutural de alta no longo prazo. O alinhamento das médias (MA50 sobrepondo a MA200) sinaliza momentos recorrentes de *Golden Cross*, caracterizando apetite comprador e confiança nas entregas operacionais da empresa.

### Resposta 2: Análise de Volume e Picos de Liquidez
Os picos significativos no volume diário de negociação concentram-se em **janelas pós-divulgação de balanços trimestrais (RI)** e rebalanceamentos de carteiras institucionais. Esse comportamento evidencia que oscilações de preço pontuais não são motivadas por especulações desordenadas, mas sim por reajustes de valor intrínseco com respaldo fundamentalista.

### Resposta 3: Estabilidade Operacional e Saúde Financeira
A consistência no padrão das oscilações da WEG reflete sua forte saúde financeira:
* **Baixa Dependência de Alavancagem Excessiva:** Protege a empresa em ciclos de juros elevados.
* **Diversificação Geográfica:** Atua como um *hedge* natural contra flutuações do cenário macroeconômico doméstico.
